Gamma Correction

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = cv2.imread("./misc/lena_color.tiff",0)
#img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img,cmap="gray")
plt.show()
```

```python
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.plot(x,y)
plt.show()
```

```python
img_norm=img/255.0
img_temp=np.power(img_norm,2.5)
img_gamma=(img_temp * 255).astype(np.uint8)
```

```python
plt.imshow(img_gamma,cmap="gray")
plt.title("Gamma Correction")
plt.show()
```

Histogram Equalization

```python
img = cv2.imread("./misc/prac1.tiff",0)
#img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img,cmap="gray")
plt.show()
```

```python
hist=[]
prob=[]
c_prob=[]
p=0
grayscale=[i for i in range(0,256)]
```

```python
pixels=list(img.flatten())
```

```python
for inten in grayscale:
    freq =pixels.count(inten)
    hist.append(freq)
    prob.append(freq/img.size)
    p=p+freq/img.size
    c_prob.append(p)
    
```

```python
plt.plot(grayscale,c_prob)
plt.title("histogram equilization")
```

```python
s=[]
```

```python
for i in range(len(grayscale)):
    s.append(round(c_prob[i]*255))
```

```python
img.shape
```

```python
new_img=np.zeros(img.shape)
print(new_img)
```

```python
for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        new_img[i,j]=s[img[i,j]]
```

```python
plt.imshow(new_img,cmap="gray")
plt.show()
```
