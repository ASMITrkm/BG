# salt and pepper noise

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
img=cv2.imread("E:/BG/lena.jpeg",0)
```

```python
def saltpepper(img,d):
    location=[index for index in np.ndindex(img.shape)]
    d=0.01
    npix_pep=npix_salt=int(img.size*d)
    sample_salt=random.sample(location,npix_salt)
    sample_pep=random.sample(location,npix_pep)
    sp_img=img.copy()
    for index in sample_salt:
        sp_img[index]=255
    for index in sample_pep:
        sp_img[index]=0
    return sp_img
    
    
```

```python
sp_img=saltpepper(img,0.2)
plt.imshow(sp_img,cmap='gray')
plt.show()
```
