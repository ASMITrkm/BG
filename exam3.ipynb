3.	Implement Object detection using HSV color space with image and video.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

Object Detection Using HSV on Image

```python
img=cv2.imread('E:\\BG\\smarties.jpg')
hsv=cv2.cvtColor(img,cv2.COLOR_BGR2HSV)
l_b=np.array([80,50,50])
u_b=np.array([140,255,255])
mask=cv2.inRange(hsv,l_b,u_b)
res=cv2.bitwise_and(img,img,mask=mask)
plt.figure(figsize=(8,4))
plt.subplot(1,4,1)
plt.imshow(img)
plt.title('Original')
plt.subplot(1,4,2)
plt.imshow(hsv)
plt.title('HSV')
plt.subplot(1,4,3)
plt.imshow(mask)
plt.title('Mask')
plt.subplot(1,4,4)
plt.imshow(res)
plt.title('Result')
plt.show()
```

Object Detection Using HSV on Video

```python
cap=cv2.VideoCapture("E:\\BG\\vid.mp4")
while True:
    ret,frame=cap.read()
    frame=cv2.resize(frame,(600,400))
    hsv=cv2.cvtColor(frame,cv2.COLOR_BGR2HSV)
    l_b=np.array([5,150,50])
    u_b=np.array([20,255,255])
    mask=cv2.inRange(hsv,l_b,u_b)
    res=cv2.bitwise_and(frame,frame,mask=mask)
    cv2.imshow('Original Video',frame)
    cv2.imshow('Mask',mask)
    cv2.imshow('Result',res)
    k=cv2.waitKey(5) & 0xFF
    if k==27:
        break
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

def nothing(x):
    pass

cv2.namedWindow('image')
cv2.createTrackbar("test","image",50,500,nothing)
cv2.createTrackbar("color/gray",'image',0,1,nothing)


while True:
    im=cv2.imread('E:\\BG\\lena.jpeg')
    im=cv2.resize(im,(600,600))
    cv2.imshow('image',im)
    test=cv2.getTrackbarPos("test","image")
    font=cv2.FONT_HERSHEY_COMPLEX
    cv2.putText(im,str(test),(50,150),font,4,(255,255,255),2)
    s=cv2.getTrackbarPos("color/gray",'image')
    if s==0:
        pass
    else:
        im=cv2.cvtColor(im,cv2.COLOR_BGR2GRAY)
    cv2.imshow('image',im)
    k=cv2.waitKey(1) & 0xFF
    if k==27:
        break
    
cv2.destroyAllWindows()
```
