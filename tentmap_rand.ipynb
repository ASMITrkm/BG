Tent Map Randomization

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img=cv2.imread('lena.jpeg',0)
plt.imshow(img,cmap='gray')
```

```python
def generate_rn(img):
    x=[]
    h=1.4
    x.append(0.3)
    for i in range(1,img.shape[0]**2):
        if x[i-1]<0.5:
            x_new=x[i-1]*h
        else:
            x_new=(1-x[i-1])*h
        x.append(x_new)
    return x
```

```python
rn = generate_rn(img)
sorted_rn=rn.copy()
sorted_rn.sort()
indices=[None]*(img.shape[0]**2)
```

```python
for i in range(len(rn)):
    for j in range(len(sorted_rn)):
        if rn[i] == sorted_rn[j]:
            indices[i] = j
            break
```

```python
img_cpy = img.copy()
img_cpy = img_cpy.flatten()

new_img = np.zeros(img.shape[0]**2)

index = 0
for i in indices:
    new_img[i] = img_cpy[index]
    index += 1
```

```python
new_img=new_img.reshape(img.shape[0],img.shape[1])
plt.imshow(new_img,cmap='gray')
plt.axis('off')
plt.show()
```
