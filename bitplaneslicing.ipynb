Bit Plane Slicing

```python
import cv2
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt 
```

```python
img=cv2.imread("E:/BG/pic/ww2.tif",0)
```

```python
list=[]
```

```python
for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        list.append(np.binary_repr(img[i][j], width=8))
```

```python
arr=np.array(list)
```

```python
bit7 = ((img >> 7) & 1) * 255
bit6 = ((img >> 6) & 1) * 255
bit5 = ((img >> 5) & 1) * 255
bit4 = ((img >> 4) & 1) * 255
bit3 = ((img >> 3) & 1) * 255
bit2 = ((img >> 2) & 1) * 255
bit1 = ((img >> 1) & 1) * 255
bit0 = (img & 1) * 255
```

```python
titles = ['Original','Bit7','Bit6','Bit5','Bit4','Bit3','Bit2','Bit1','Bit0']
images = [img, bit7, bit6, bit5, bit4, bit3, bit2, bit1, bit0]

plt.figure(figsize=(10,8))
for i in range(9):
    plt.subplot(3,3,i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.show()
```
