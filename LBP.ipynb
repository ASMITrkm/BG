# Local Binary Pattern

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
def cal_lbp(img,x,y):
    centre=img[x,y]
    arr=[]

    arr.append(1 if img[x-1,y-1]>=centre else 0)
    arr.append(1 if img[x-1,y]>=centre else 0)
    arr.append(1 if img[x-1,y+1]>=centre else 0)
    arr.append(1 if img[x,y+1]>=centre else 0)
    arr.append(1 if img[x+1,y+1]>=centre else 0)
    arr.append(1 if img[x+1,y]>=centre else 0)
    arr.append(1 if img[x+1,y-1]>=centre else 0)
    arr.append(1 if img[x,y-1]>=centre else 0)

    power=[1,2,4,8,16,32,64,128]
    value=0
    for i in range(0,8):
        value+=arr[i]*power[i]
    return value    
```

```python
img=cv2.imread("E:/BG/lena.jpeg",0)
```

```python
h,w=img.shape
lbp=np.zeros((h,w),dtype=np.uint8)
for i in range(1,h-1):
    for j in range(1,w-1):
        lbp[i,j]=cal_lbp(img,i,j)
plt.imshow(lbp,cmap="gray")
plt.title("LBP image")
plt.axis("off")
plt.show()
        
```
