4.	Implement Histogram equalization on an image (without a library function). Now, use it to equalize every frame of a video.

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img=cv2.imread('E:\\BG\\bird.jpeg',0)
```

```python


grayscale=range(256)

def histogram(image):
 pixels=img.flatten().tolist()
 hist=[] 
 prob=[]
 c_prob=[]
 
 p=0
 for i in grayscale:
    freq=pixels.count(i)
    hist.append(freq)
    probability=freq/len(pixels)
    prob.append(probability)
    p+=probability
    c_prob.append(p)
 return hist,prob,c_prob
hist,prob,c_prob=histogram(img)
plt.bar(grayscale,hist)
plt.xlabel('Grayscale')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.show()
```

```python
cap=cv2.VideoCapture("E:\\BG\\vid.mp4")
while True:
    ret,frame=cap.read()
    if not ret:
        break
    frame=cv2.resize(frame,(700,500))
    gray=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    hist,prob,c_prob=histogram(gray)
    plt.bar(grayscale,hist)
    plt.xlabel('Grayscale')
    plt.ylabel('Frequency')
    plt.title('Histogram')
    plt.show()
    cv2.imshow('Original Video',frame)
    cv2.imshow('Grayscale Video',gray)
    k=cv2.waitKey(30) & 0xFF
    if k==27:
        break
    cap.release()
    cv2.destroyAllWindows()
    
```
